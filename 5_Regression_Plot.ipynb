# lmplot()
*regression plot is used in machine learning so we will not fully discuss what, how and where it is used
but we will see how it is created and what are the values we can put inside it*

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
tips = sns.load_dataset('tips')
```

```python
tips
```

```python
sns.lmplot(x = 'total_bill', y = 'tip', data = tips) #Linera Regressio origin
```

```python
sns.lmplot(x = 'total_bill', y = 'tip', data = tips, hue = 'sex')
```

```python
sns.lmplot(x = 'total_bill', y = 'tip', data = tips, hue = 'sex', palette = 'pink')
```

```python
sns.lmplot(x = 'total_bill', y = 'tip', data = tips, hue = 'sex', palette = 'pink', markers = ['o','v'])
```

```python
sns.lmplot(x = 'total_bill', y = 'tip', data = tips, hue = 'sex', palette = 'pink', markers = ['o','v'], scatter_kws= {'s':100})
```
