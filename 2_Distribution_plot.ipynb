```python
import seaborn as sns
import matplotlib.pyplot as plt
```

We use distribution plots to visualize the distribution of **quantitative data**.


Let's discuss some plots that allow us to visualize the distribution of a data set. These plots are:

* distplot/histplot
* jointplot
* pairplot
* rugplot

```python
df = sns.load_dataset('tips') #already available dataset on the official site of seaborn
df
```

**QUANTITATIVE DATA-**
total_bill, tip

**Distribution Plot**
Syntax - data[''].unique()

```python
df['size'].unique() #Size is categorical or qualitative data
```

```python
#distribution plot
sns.displot(df['total_bill']) #displot ot histplot
```

```python
sns.displot(df['total_bill'], bins = 50)
```

```python
sns.histplot(df['total_bill'], bins = 50)
```

```python
#Kernel destination estimation - show skewness
sns.histplot(df['total_bill'], bins = 50, kde = True)
```

```python
plt.subplot(1,2,1)
sns.histplot(df['total_bill'], bins = 20, kde = True)
plt.subplot(1,2,2)
sns.histplot(df['tip'], bins = 20, kde = True)
```

**Joint Plot** Syntax - sns.jointplot(x = 'data name for row', y = 'data name for column', data ='data', kind = ' scatter') kind is bydefault

```python
sns.jointplot(x = 'total_bill', y = 'tip',data = df, kind = 'scatter') 
```

```python
sns.jointplot(x = 'total_bill', y = 'tip',data = df, kind = 'hex') 
```

```python
sns.jointplot(x = 'total_bill', y = 'tip',data = df, kind = 'kde') 
```

```python
sns.jointplot(x = 'total_bill', y = 'tip',data = df, kind = 'reg') 
```

```python
sns.jointplot(x = 'total_bill', y = 'tip',data = df, kind = 'resid') 
```

**Pair Plot**

```python
sns.pairplot(df) # compare numerical column - creting pair
```

```python
sns.pairplot(df, hue = 'sex') #hue--. dividing based on category
```

```python
sns.pairplot(df, hue = 'size', palette = 'rainbow') #palette --> color
```

**Rug Plot** - a kind of doormet

```python
sns.rugplot(df['total_bill'])
```
